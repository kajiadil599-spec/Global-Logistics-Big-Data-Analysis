```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Project: Global Logistics & Supply Chain Optimization
# Scale: 50,000 Rows

def generate_big_logistics_data(rows=50000):
    np.random.seed(42)

    # 1. Base Dimensions
    regions = ['North America', 'Europe', 'Asia-Pacific', 'Middle East', 'Latin America']
    categories = ['Electronics', 'Fashion', 'Automotive', 'Healthcare', 'Home Decor']
    ship_modes = ['Air Priority', 'Ocean Freight', 'Land Transport']
    status = ['Delivered', 'In Transit', 'Delayed', 'Cancelled']

    # 2. Generating Random Data
    data = {
        'Order_ID': [f'ORD-{100000 + i}' for i in range(rows)],
        'Order_Date': [datetime(2025, 1, 1) + timedelta(days=np.random.randint(0, 365)) for i in range(rows)],
        'Region': np.random.choice(regions, rows),
        'Product_Category': np.random.choice(categories, rows),
        'Ship_Mode': np.random.choice(ship_modes, rows),
        'Unit_Price': np.random.uniform(50, 2000, rows).round(2),
        'Quantity': np.random.randint(1, 15, rows),
        'Shipping_Cost': np.random.uniform(10, 500, rows).round(2),
        'Delivery_Status': np.random.choice(status, rows, p=[0.7, 0.15, 0.1, 0.05])
    }

    df = pd.DataFrame(data)

    # 3. Feature Engineering (The "Data Analyst" touch)
    df['Total_Sales'] = (df['Unit_Price'] * df['Quantity']).round(2)
    df['Profit_Margin'] = np.random.uniform(0.05, 0.25, rows).round(4)
    df['Total_Profit'] = (df['Total_Sales'] * df['Profit_Margin']).round(2)

    # Add logic for Delay Days (Calculated column)
    df['Estimated_Lead_Time'] = np.random.randint(3, 15, rows)
    df['Actual_Delay_Days'] = df['Delivery_Status'].apply(lambda x: np.random.randint(1, 10) if x == 'Delayed' else 0)

    return df

# Create and save
big_df = generate_big_logistics_data()
big_df.to_csv('Global_Logistics_Master.csv', index=False)
print("Big dataset successfully generated with 50,000 rows!")
```
